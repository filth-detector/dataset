```python
import os
from PIL import Image

def generate_blank_rgb_masks(input_img_dir, output_mask_dir):
    os.makedirs(output_mask_dir, exist_ok=True)
    
    valid_extensions = ('.png', '.jpg', '.jpeg')
    files = [f for f in os.listdir(input_img_dir) if f.lower().endswith(valid_extensions)]
    
    if not files:
        print(f"No images found in {input_img_dir}")
        return

    print(f"Found {len(files)} images.")
    
    success_count = 0
    
    for filename in files:
        img_path = os.path.join(input_img_dir, filename)
        
        with Image.open(img_path) as img:
            width, height = img.size
                
        blank_mask = Image.new("RGB", (width, height), color=(0, 0, 0))
            
        name_without_ext = os.path.splitext(filename)[0]
        mask_filename = f"{name_without_ext}.png"
        mask_path = os.path.join(output_mask_dir, mask_filename)
            
        blank_mask.save(mask_path)
        success_count += 1

    print(f"Generated {success_count} blank RGB masks in '{output_mask_dir}'")
```

```python

IMAGE_DIR = "./data/final/clean/JPEGImages"
MASK_DIR  = "./data/final/clean/SegmentationClass"
    
generate_blank_rgb_masks(IMAGE_DIR, MASK_DIR)
```

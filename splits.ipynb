```python
import os
import random
import shutil

def create_splits(images_dir, masks_dir, output_dir, train_ratio):
    image_files = [f for f in os.listdir(images_dir) if os.path.isfile(os.path.join(images_dir, f))]
    mask_files = [f for f in os.listdir(masks_dir) if os.path.isfile(os.path.join(masks_dir, f))]

    image_map = {os.path.splitext(f)[0]: f for f in image_files}
    mask_map = {os.path.splitext(f)[0]: f for f in mask_files}

    valid_bases = sorted(list(set(image_map.keys()).intersection(set(mask_map.keys()))))

    random.seed(42)
    random.shuffle(valid_bases)

    split_idx = int(len(valid_bases) * train_ratio)
    
    splits = {
        "train": valid_bases[:split_idx],
        "test": valid_bases[split_idx:]
    }

    for split_name, bases in splits.items():
        out_images_dir = os.path.join(output_dir, split_name, "images")
        out_masks_dir = os.path.join(output_dir, split_name, "masks")
        
        os.makedirs(out_images_dir, exist_ok=True)
        os.makedirs(out_masks_dir, exist_ok=True)

        for base in bases:
            shutil.copy2(os.path.join(images_dir, image_map[base]), os.path.join(out_images_dir, image_map[base]))
            shutil.copy2(os.path.join(masks_dir, mask_map[base]), os.path.join(out_masks_dir, mask_map[base]))
```

```python
create_splits("data/final/meat/JPEGImages", "data/final/meat/SegmentationClass", "data/final/meat-splits", train_ratio=0.85)
```

```python
create_splits("data/final/vegetables/JPEGImages", "data/final/vegetables/SegmentationClass", "data/final/vegetables-splits", train_ratio=0.85)
```

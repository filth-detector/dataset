```python
import os
import hashlib

def calculate_md5(filepath):
    hasher = hashlib.md5()
    try:
        with open(filepath, 'rb') as f:
            chunk = f.read(8192)
            while chunk:
                hasher.update(chunk)
                chunk = f.read(8192)
        return hasher.hexdigest()
    except OSError as e:
        print(f"Error reading {filepath}: {e}")
        return None

def remove_identical_files(directory_path):
    unique_hashes = {}
    
    deleted_count = 0
    
    print(f"scanning directory: {directory_path} ...")

    for filename in sorted(os.listdir(directory_path)):
        filepath = os.path.join(directory_path, filename)
        
        if os.path.isfile(filepath):
            file_hash = calculate_md5(filepath)
            
            if file_hash:
                if file_hash in unique_hashes:
                    original = unique_hashes[file_hash]
                    print(f"Deleting duplicate: {filename} (Identical to {original})")
                    
                    try:
                        os.remove(filepath)
                        deleted_count += 1
                    except OSError as e:
                        print(f"Error deleting {filename}: {e}")
                else:
                    unique_hashes[file_hash] = filename

    print(f"\nDone! Cleaned up {deleted_count} duplicate files.")

if __name__ == "__main__":
    target_folder = r"/home/ugnius/Documents/KTU/IV-kursas/II-semestras/BBP/V2/dataset/data/sorted/clean"

    
    if os.path.exists(target_folder):
        remove_identical_files(target_folder)
    else:
        print("Folder not found. Please check the path.")
```

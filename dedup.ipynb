```python
import os
import hashlib

def calculate_md5(filepath):
    hasher = hashlib.md5()
    try:
        with open(filepath, 'rb') as f:
            # Read the file in small chunks
            chunk = f.read(8192)
            while chunk:
                hasher.update(chunk)
                chunk = f.read(8192)
        return hasher.hexdigest()
    except OSError as e:
        print(f"Error reading {filepath}: {e}")
        return None

def remove_identical_files(directory_path):
    # Dictionary to store hash -> first_filename
    unique_hashes = {}
    
    # Counter for stats
    deleted_count = 0
    
    print(f"scanning directory: {directory_path} ...")
    
    # List all files in the directory
    # sorted() ensures we process files in a deterministic order (e.g., image_01 before image_02)
    for filename in sorted(os.listdir(directory_path)):
        filepath = os.path.join(directory_path, filename)
        
        # Check if it's actually a file (and not a subdirectory)
        if os.path.isfile(filepath):
            file_hash = calculate_md5(filepath)
            
            if file_hash:
                if file_hash in unique_hashes:
                    # We have seen this hash before! It's a duplicate.
                    original = unique_hashes[file_hash]
                    print(f"❌ Deleting duplicate: {filename} (Identical to {original})")
                    
                    try:
                        os.remove(filepath)
                        deleted_count += 1
                    except OSError as e:
                        print(f"Error deleting {filename}: {e}")
                else:
                    # First time seeing this hash. Keep the file.
                    unique_hashes[file_hash] = filename

    print(f"\nDone! Cleaned up {deleted_count} duplicate files.")

if __name__ == "__main__":
    target_folder = r"/home/ugnius/Documents/KTU/IV-kursas/II-semestras/BBP/V2/dataset/data/sorted/vegetables/vegetables1"

    
    if os.path.exists(target_folder):
        remove_identical_files(target_folder)
    else:
        print("Folder not found. Please check the path.")
```
